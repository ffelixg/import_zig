Create the project folder inside the current directory

```python
from pathlib import Path
project_folder = Path("my_zig_extension")
assert not project_folder.exists()
project_folder.mkdir()
```

Prepare the project folder for development

```python
import import_zig
import_zig.prepare(project_folder, "my_module_name", "my_root_source_file_name.zig")
```

Simulate development. You should have ZLS support inside `my_root_source_file_name.zig`

```python
with open(project_folder / "my_root_source_file_name.zig", "w") as f:
    f.write("""\
const std = @import("std");
const c = @import("c");

pub fn hey() !*c.PyObject {
    const str = "hey";
    return c.PyUnicode_FromStringAndSize(str, str.len) orelse return error.PyErr;
}
""")

# you may also want to ignore the build.zig files
with open(project_folder / ".gitignore", "w") as f:
    f.writelines(["build.zig\n", "build.zig.zon\n"])

with open(project_folder / "__init__.py", "w") as f:
    pass
```

Compile and use the extension

```python
import_zig.compile_prepared(project_folder, project_folder)
from my_zig_extension import my_module_name
print(my_module_name.hey())
```
